# Preprocessing

```python
import warnings
import logging
import pandas as pd
import kagglehub
import os

# Hide Python warnings
warnings.filterwarnings("ignore")

# Suppress Jupyter client warnings specifically
logging.getLogger("jupyter_client").setLevel(logging.ERROR)

# 1. Load the dataset
try:
    path = kagglehub.dataset_download("aliiihussain/amazon-sales-dataset")
    csv_path = os.path.join(path, "amazon_sales_dataset.csv")
    df = pd.read_csv(csv_path)
    print("Dataset loaded successfully.\n")
except FileNotFoundError:
    print("Error: 'amazon_sales_dataset.csv' not found.")


# Quick check for data issues before mining
print(f"Missing values:\n{df.isnull().sum()}")
print(f"Duplicate rows: {df.duplicated().sum()}")
# Basic noise check: ensure no negative quantities or impossible ratings
print(f"Invalid Ratings Check: {df[df['rating'] > 5].shape[0]} rows")
```

## Data Preprocessing

Before applying the analysis methods, the dataset was carefully reviewed to ensure its quality and suitability for data mining techniques.

### Data Quality Checks

The dataset was found to be well-structured and relatively clean. Specifically:

- No missing values were observed in the key variables used for analysis  
- No duplicate rows were found  
- All rating values were within valid bounds (no values exceeding expected limits)  

Because of this, no major data cleaning or imputation was required. This allowed the analysis to proceed directly without introducing bias from data repair steps.

---

#Feature Transformation

```python
# --- DATA PREPARATION (Discretization) ---
# We must bin continuous data for Association/Frequent Itemset mining.
df['disc_bin'] = pd.cut(df['discount_percent'], bins=[-1, 5, 20, 31], labels=['Low_Disc', 'Mid_Disc', 'High_Disc'])
df['rate_bin'] = pd.cut(df['rating'], bins=[0, 2, 4, 5.1], labels=['Poor_Rate', 'Avg_Rate', 'High_Rate'])
df['qty_bin'] = pd.cut(df['quantity_sold'], bins=[0, 1, 3, 6], labels=['Single_Qty', 'Multi_Qty', 'Bulk_Qty'])
df['review_bin'] = pd.cut(df['review_count'], bins=[-1, 100, 300, 501], labels=['Niche', 'Steady', 'Popular'])
```



### Feature Transformation

Although the dataset was clean, several variables needed to be transformed to support the chosen analysis methods.

Association Rule Mining and categorical analysis require discrete variables rather than continuous values. Therefore, the following variables were discretized into meaningful tiers:

| Original Variable | Transformed Variable | Categories |
|---|---|---|
| Discount (%) | `disc_bin` | Low / Mid / High |
| Rating | `rate_bin` | Poor / Average / High |
| Quantity Sold | `qty_bin` | Single / Multi / Bulk |
| Review Count | `review_bin` | Niche / Steady / Popular |

This transformation enables:

- more interpretable pattern discovery  
- compatibility with Apriori-based rule mining  
- consistent comparison across product categories
